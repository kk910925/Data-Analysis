```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Python商業分析/coffee_shop.csv')
print(type(df))
```

```python
df.head()
```

```python
df[['Date','Price','Quantity']].head()
df['Total Price'] = df['Price'] * df['Quantity']
df.head()
```

```python
groups = df.groupby('Product')
print(type(groups),len(groups))
```

```python
for Product_name,group_data in df.groupby('Product'):
  print(f'產品名:{Product_name},共{len(group_data)}項')
  display(group_data.head(n=10))
  print()
```

```python
group_product_latte = df.groupby('Product').get_group('Latte')
print(f'共{len(group_product_latte)}項')
display(group_product_latte)
```

```python
#第一種寫法
group_payment_creditcard = df.groupby('Payment').get_group('Credit Card')
print(f'共{len(group_payment_creditcard)}項')
display(group_payment_creditcard)
```

```python
#第二種寫法
member_latte = df[(df['Product']=='Latte')&(df['Member']=='Yes')]
display(member_latte)
```

```python
for product_name,group_data in df.groupby('Product'):
  product_quantity = group_data['Quantity'].sum()
  item_price = group_data['Price'].iloc[0]
  print(f'總銷售量:{product_quantity},總銷售額:{product_quantity * item_price}')
```

```python
#產品銷售量
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style('whitegrid')
product_sales = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
#繪製長條圖
plt.figure(figsize=(10, 6))
sns.barplot(x=product_sales.index, y=product_sales.values, palette='viridis')
plt.title('Amount')
plt.xlabel('product_name')
plt.ylabel('sale_quantity')
plt.xticks(rotation=45)
plt.savefig('product_sales.png')
plt.show()
```

```python
#銷售總額(Category)
df['Total Price'] = df['Price'] * df['Quantity']
Category_price = df.groupby('Category')['Total Price'].sum()
# 繪製銷售總額圓餅圖
plt.figure(figsize=(8, 8))
plt.pie(Category_price.values, labels=Category_price.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'])
plt.title('Category_price')
plt.axis('equal')  # 使圓餅圖比例相等
plt.show()
```

```python
#銷售總額(Product)
df['Total Price'] = df['Price'] * df['Quantity']
Product_price = df.groupby('Product')['Total Price'].sum()
# 繪製銷售總額圓餅圖
plt.figure(figsize=(8, 8))
plt.pie(Product_price.values, labels=Product_price.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'])
plt.title('Product_price')
plt.axis('equal')  # 使圓餅圖比例相等
plt.show()
```

```python
#計算每個會員類別中，各商品類別的總銷售額
print('Groupby and sum')
member_category_price = df.groupby(['Member','Category'])['Total Price'].sum()
display(member_category_price)
```

```python
print('Unstack')
member_category_price = member_category_price.unstack()
display(member_category_price)
```

```python
print('Reset Index')
#繪製堆疊長條圖或分組長條圖
#將DataFrame轉換為適合繪圖的格式
member_category_price = member_category_price.reset_index()
display(member_category_price)
```
