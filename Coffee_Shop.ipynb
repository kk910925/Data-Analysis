```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Python商業分析/coffee_shop.csv')
print(type(df))
```

```python
df.head()
```

```python
df[['Date','Price','Quantity']].head()
df['Total Price'] = df['Price'] * df['Quantity']
df.head()
```

```python
groups = df.groupby('Product')
print(type(groups),len(groups))
```

```python
for Product_name,group_data in df.groupby('Product'):
  print(f'產品名:{Product_name},共{len(group_data)}項')
  display(group_data.head(n=10))
  print()
```

```python
group_product_latte = df.groupby('Product').get_group('Latte')
print(f'共{len(group_product_latte)}項')
display(group_product_latte)
```

```python
#第一種寫法
group_payment_creditcard = df.groupby('Payment').get_group('Credit Card')
print(f'共{len(group_payment_creditcard)}項')
display(group_payment_creditcard)
```

```python
#第二種寫法
member_latte = df[(df['Product']=='Latte')&(df['Member']=='Yes')]
display(member_latte)
```

```python
for product_name,group_data in df.groupby('Product'):
  product_quantity = group_data['Quantity'].sum()
  item_price = group_data['Price'].iloc[0]
  print(f'總銷售量:{product_quantity},總銷售額:{product_quantity * item_price}')
```
